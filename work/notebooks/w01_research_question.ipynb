## 2. The question: decision, action, cost of a wrong call

### Research question
How can observable search, content, freshness, and engagement signals help prioritize which pages should be reviewed first for a possible content refresh?

### Unit of analysis
The unit of analysis is one content page, represented by one pseudonymized `content_id` and its performance measurements.

### Decision and action

The expected output is a ranked list of pages with a priority score and clear reasons explaining why each page was recommended.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

# Load the anonymized starter dataset from my GitHub repository
data_url = (
    "https://raw.githubusercontent.com/"
    "Dania-Yasir/flyrak-project/main/"
    "data/raw/content_refresh_anonymized.csv"
)

df = pd.read_csv(data_url)

# 1. Total number of pages
total_pages = len(df)

# 2. Percentage of pages with a declining trend
declining_rate = (
    df["trend_direction"]
    .astype(str)
    .str.lower()
    .eq("down")
    .mean() * 100
)

# 3. Pages that are both stale and still visible
stale_visible_pages = (
    (pd.to_numeric(df["days_since_last_update"], errors="coerce") >= 180)
    & (pd.to_numeric(df["impressions_90d"], errors="coerce") >= 500)
).sum()

print(f"Total pages in the starter dataset: {total_pages:,}")
print(f"Pages with a declining trend: {declining_rate:.1f}%")
print(f"Stale and visible pages: {stale_visible_pages:,}")
```

### What these numbers suggest

The starter dataset contains 30,000 content pages, which is too many for a reviewer to inspect manually without prioritization. About 54.2% of the pages have a declining trend in the current data window, showing that many pages could potentially require attention. However, only 17 pages meet the simple `stale and visible` rule used here: at least 180 days since the last update and at least 500 impressions in the last 90 days.

These observed numbers support the need for a ranked review system. A single rule may find some useful candidates, but combining multiple observable signals could help reviewers decide which pages deserve attention first.
